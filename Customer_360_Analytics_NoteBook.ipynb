```python
# 1. POS se customer-level summary
pos_summary = pos_clean.groupby('customer_id').agg(
    total_transactions=('invoice_no', 'count'),
    total_spend=('total_amount', 'sum'),
    avg_order_value=('total_amount', 'mean'),
    total_cancelled=('is_cancelled', 'sum')
).reset_index()

# 2. Website se customer-level summary
website_summary = website_clean.groupby('customer_id').agg(
    total_visits=('session_id', 'count'),
    avg_session_duration=('session_duration_sec', 'mean'),
    total_conversions=('converted', 'sum')
).reset_index()

# 3. Support se customer-level summary
support_summary = support.groupby('customer_id').agg(
    total_tickets=('ticket_id', 'count'),
    avg_resolution_time=('resolution_time_hours', 'mean'),
    avg_satisfaction_score=('satisfaction_score', 'mean')
).reset_index()

# 4. Email se customer-level summary
email_summary = email.groupby('customer_id').agg(
    total_emails_received=('email_id', 'count'),
    total_opened=('opened', 'sum'),
    total_clicked=('clicked', 'sum'),
    total_converted_to_sale=('converted_to_sale', 'sum')
).reset_index()

# 5. Payment se customer-level summary
payment_summary = payment.groupby('customer_id').agg(
    total_payments=('payment_id', 'count'),
    total_amount_paid=('amount_paid', 'sum')
).reset_index()

# Check karein sab sahi bana
print(pos_summary.head())
print(website_summary.head())
print(support_summary.head())
print(email_summary.head())
print(payment_summary.head())
```

```python
# Sab summaries ko customer_id par merge karo (left join se koi customer miss nahi hoga)

customer_360 = pos_summary.merge(website_summary, on='customer_id', how='left') \
                           .merge(support_summary, on='customer_id', how='left') \
                           .merge(email_summary, on='customer_id', how='left') \
                           .merge(payment_summary, on='customer_id', how='left')

# Shape check karo - kitne customers aur kitne columns bane
print(customer_360.shape)
customer_360.head()
```

```python
# Missing values ko sensible tareeke se fill karo
count_cols = ['total_transactions','total_visits','total_conversions',
              'total_tickets','total_emails_received','total_opened',
              'total_clicked','total_converted_to_sale','total_payments']

for col in count_cols:
    if col in customer_360.columns:
        customer_360[col] = customer_360[col].fillna(0)

# Spend/amount jaisi columns bhi 0 se fill ho sakti hain (matlab kuch nahi khareeda)
amount_cols = ['total_spend','total_amount_paid']
for col in amount_cols:
    if col in customer_360.columns:
        customer_360[col] = customer_360[col].fillna(0)
```

```python
print("Duplicate customer_ids:", customer_360['customer_id'].duplicated().sum())
```

```python
# 1. payment_date ko datetime format mein convert karo
payment['payment_date'] = pd.to_datetime(payment['payment_date'])

# 2. Sirf successful payments lena behtar hai (Failed/Pending exclude)
successful_payments = payment[payment['payment_status'] == 'Success']

# 3. Har customer ka last (sabse recent) payment date nikalo
last_purchase = successful_payments.groupby('customer_id')['payment_date'].max().reset_index()
last_purchase.rename(columns={'payment_date': 'last_purchase_date'}, inplace=True)

# 4. Snapshot date se Recency (days) nikalo
snapshot_date = payment['payment_date'].max() + pd.Timedelta(days=1)
last_purchase['recency_days'] = (snapshot_date - last_purchase['last_purchase_date']).dt.days

print(last_purchase.head())
```

```python
# Recency ko customer_360 mein merge karo
customer_360 = customer_360.merge(
    last_purchase[['customer_id', 'recency_days']],
    on='customer_id',
    how='left'
)

# Jinka recency_days NaN hai (kabhi successful payment nahi kiya), unko max recency de do
customer_360['recency_days'] = customer_360['recency_days'].fillna(customer_360['recency_days'].max())

print(customer_360[['customer_id', 'recency_days', 'total_transactions', 'total_spend']].head())
```

```python
# RFM Score (1-5 scale) banate hain using quantile-based ranking

# Recency: kam days = zyada score (isliye labels reverse)
customer_360['R_score'] = pd.qcut(customer_360['recency_days'], 5, labels=[5,4,3,2,1])

# Frequency: zyada transactions = zyada score
customer_360['F_score'] = pd.qcut(customer_360['total_transactions'].rank(method='first'), 5, labels=[1,2,3,4,5])

# Monetary: zyada spend = zyada score
customer_360['M_score'] = pd.qcut(customer_360['total_spend'], 5, labels=[1,2,3,4,5])

# Teeno scores ko combine karke ek RFM segment string banao
customer_360['RFM_Score'] = (
    customer_360['R_score'].astype(str) +
    customer_360['F_score'].astype(str) +
    customer_360['M_score'].astype(str)
)

print(customer_360[['customer_id', 'recency_days', 'total_transactions', 'total_spend',
                     'R_score', 'F_score', 'M_score', 'RFM_Score']].head(10))
```

```python
def rfm_segment(row):
    R, F, M = int(row['R_score']), int(row['F_score']), int(row['M_score'])
    
    if R >= 4 and F >= 4 and M >= 4:
        return 'Champions'
    elif R >= 4 and F >= 3:
        return 'Loyal Customers'
    elif R >= 4 and F <= 2:
        return 'New Customers'
    elif R == 3:
        return 'Potential Loyalist'
    elif R <= 2 and F >= 4:
        return 'At Risk'
    elif R <= 2 and F <= 2 and M <= 2:
        return 'Lost'
    else:
        return 'Needs Attention'

customer_360['Customer_Segment'] = customer_360.apply(rfm_segment, axis=1)

# Check karo har segment mein kitne customers aaye
print(customer_360['Customer_Segment'].value_counts())

# Sample dekho
print(customer_360[['customer_id', 'R_score', 'F_score', 'M_score', 'Customer_Segment']].head(10))
```

```python
# Simple CLV = Average Order Value * Total Transactions
# (Ye "historical CLV" hai - jitna customer ne ab tak company ko diya)
customer_360['CLV'] = customer_360['avg_order_value'] * customer_360['total_transactions']

print(customer_360[['customer_id', 'avg_order_value', 'total_transactions', 'CLV']].head(10))
```

```python
# ================================
# 1. Repeat Purchase Rate
# ================================
# Repeat customer = jisne 1 se zyada transaction kiya (matlab wapis aaya)
repeat_customers = customer_360[customer_360['total_transactions'] > 1].shape[0]
total_customers = customer_360.shape[0]

repeat_purchase_rate = (repeat_customers / total_customers) * 100
print(f"Total Customers: {total_customers}")
print(f"Repeat Customers: {repeat_customers}")
print(f"Repeat Purchase Rate: {repeat_purchase_rate:.2f}%")

print("\n" + "="*50 + "\n")

# ================================
# 2. Churn Flag banao (customer-level)
# ================================
# Agar customer ne 60 din se zyada koi purchase nahi ki -> Churned mano
churn_threshold = 60  # din - is number ko aap apni marzi se adjust kar sakte ho

customer_360['Churn_Flag'] = customer_360['recency_days'].apply(
    lambda x: 'Churned' if x > churn_threshold else 'Active'
)

print(customer_360['Churn_Flag'].value_counts())

print("\n" + "="*50 + "\n")

# ================================
# 3. Overall Churn Rate aur Retention Rate
# ================================
churn_rate = (customer_360['Churn_Flag'] == 'Churned').mean() * 100
retention_rate = 100 - churn_rate

print(f"Overall Churn Rate: {churn_rate:.2f}%")
print(f"Overall Retention Rate: {retention_rate:.2f}%")

print("\n" + "="*50 + "\n")

# ================================
# 4. Sample check karo
# ================================
print(customer_360[['customer_id', 'recency_days', 'total_transactions', 'CLV', 'Churn_Flag']].head(10))
```

```python
# 1. Date ko datetime format mein convert karo
pos['invoice_date'] = pd.to_datetime(pos['invoice_date'])

# 2. Month nikaalo har transaction se
pos['month'] = pos['invoice_date'].dt.to_period('M')

# 3. Monthly revenue trend banao
monthly_revenue = pos.groupby('month')['total_amount'].sum().reset_index()
monthly_revenue['month'] = monthly_revenue['month'].astype(str)

print(monthly_revenue)
```

```python
# Har product ka total revenue nikaalo, sabse zyada bikne wale dikhao
product_performance = pos.groupby('product_name').agg(
    total_revenue=('total_amount', 'sum'),
    total_quantity_sold=('quantity', 'sum'),
    total_orders=('invoice_no', 'count')
).reset_index().sort_values('total_revenue', ascending=False)

print(product_performance.head(10))
```

```python
# Category Performance
category_performance = pos.groupby('product_category').agg(
    total_revenue=('total_amount', 'sum'),
    total_quantity_sold=('quantity', 'sum'),
    total_orders=('invoice_no', 'count')
).reset_index().sort_values('total_revenue', ascending=False)

print("Category Performance:")
print(category_performance)

print("\n" + "="*50 + "\n")

# Regional/Store Sales
store_performance = pos.groupby('store_id').agg(
    total_revenue=('total_amount', 'sum'),
    total_orders=('invoice_no', 'count')
).reset_index().sort_values('total_revenue', ascending=False)

print("Store Performance:")
print(store_performance)
```

```python
# Seasonal Analysis - kis season mein zyada bikri hoti hai
pos['season'] = pos['invoice_date'].dt.month.map({
    12: 'Winter', 1: 'Winter', 2: 'Winter',
    3: 'Spring', 4: 'Spring', 5: 'Spring',
    6: 'Summer', 7: 'Summer', 8: 'Summer',
    9: 'Fall', 10: 'Fall', 11: 'Fall'
})

seasonal_sales = pos.groupby('season')['total_amount'].sum().reset_index().sort_values('total_amount', ascending=False)
print(seasonal_sales)
```
